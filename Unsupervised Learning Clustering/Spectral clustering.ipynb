## <center> **Spectral clustering**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 

from sklearn.cluster import SpectralClustering
```

```python
pokemon = pd.read_csv('data/pokemon.csv')
X = pokemon[['Attack', 'Defense']]
#
spec_k3 = SpectralClustering(n_clusters=3, random_state=42)
spec_k4 = SpectralClustering(n_clusters=4, random_state=42)
#
spec_k3.fit(X)
spec_k4.fit(X)
#
pred_k3 = spec_k3.labels_
pred_k4 = spec_k4.labels_
#
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 7))
plot_0 = sns.scatterplot(x=pokemon.Attack, y=pokemon.Defense, c=pred_k3, ax=ax[0])
plot_1 = sns.scatterplot(x=pokemon.Attack, y=pokemon.Defense, c=pred_k4, ax=ax[1])
```

```python
house = pd.read_csv('data/housing.csv')
#
spec_k3 = SpectralClustering(n_clusters=3, random_state=42)
spec_k4 = SpectralClustering(n_clusters=4, random_state=42)
#
spec_k3.fit(house)
spec_k4.fit(house)
#
pred_k3 = spec_k3.labels_
pred_k4 = spec_k4.labels_
#
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 7))
plot_0 = sns.scatterplot(x=house.year, y=house.price, c=pred_k3, ax=ax[0])
plot_0 = sns.scatterplot(x=house.year, y=house.price, c=pred_k4, ax=ax[1])
```
