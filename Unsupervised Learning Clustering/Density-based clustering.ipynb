## <center> **Density-based clustering**

```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import DBSCAN
```

```python
pokemon = pd.read_csv('data/pokemon.csv')
house = pd.read_csv('data/housing.csv')
#
X1 = pokemon[['Attack', 'Defense']]
X2 = house
```

```python
db_pokemon_pred = DBSCAN(eps=3, min_samples=3).fit(X1)
db_house_pred = DBSCAN(eps=3, min_samples=3).fit(X2)
#
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 7))
pokemon_plot = sns.scatterplot(x=pokemon.Attack, y=pokemon.Defense, c=db_pokemon_pred.labels_, ax=ax[0])
house_plot = sns.scatterplot(x=house.year, y=house.price, c=db_house_pred.labels_, ax=ax[1])
```

```python
db_pokemon_pred = DBSCAN(eps=1, min_samples=3).fit(X1)
db_house_pred = DBSCAN(eps=1, min_samples=3).fit(X2)
#
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 7))
pokemon_plot = sns.scatterplot(x=pokemon.Attack, y=pokemon.Defense, c=db_pokemon_pred.labels_, ax=ax[0])
house_plot = sns.scatterplot(x=house.year, y=house.price, c=db_house_pred.labels_, ax=ax[1])
```

```python
db_pokemon_pred = DBSCAN(eps=0.5, min_samples=1).fit(X1)
db_house_pred = DBSCAN(eps=0.5, min_samples=1).fit(X2)
#
fig, ax = plt.subplots(nrows=2, ncols=1, figsize=(10, 7))
pokemon_plot = sns.scatterplot(x=pokemon.Attack, y=pokemon.Defense, c=db_pokemon_pred.labels_, ax=ax[0])
house_plot = sns.scatterplot(x=house.year, y=house.price, c=db_house_pred.labels_, ax=ax[1])
```

```python
db_house_pred = DBSCAN(eps=1.15, min_samples=2).fit(X2)
#
house_plot = sns.scatterplot(x=house.year, y=house.price, c=db_house_pred.labels_)
```
